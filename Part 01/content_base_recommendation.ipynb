```python
def give_rec(title, model, topk=10):
    if title not in indices:
        raise ValueError(f"Cannot find '{title}' in dataset.")

    idx = indices[title]
    scores = list(enumerate(model[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    scores = scores[1 : topk + 1]
    movie_indices = [i for i, _ in scores]
    return data.iloc[movie_indices][["original_title", "overview"]]
```
