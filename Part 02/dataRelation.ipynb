```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
rating = pd.read_parquet(
    path=r"./datasets/rating.parquet",
    engine="pyarrow",
    columns=["userId", "movieId", "rating"],
)
# rating.head()
```

```python
movie = pd.read_parquet(
    path=r"./datasets/movie.parquet",
    engine="pyarrow",
)
# movie.head()
```

```python
data = pd.merge(left=rating, right=movie, on="movieId")
```

```python
# data.head()
# data.shape
```

```python
movie_rating_count = (
    data.groupby(by="title")["rating"]
    .count()
    .reset_index()
    .rename(columns={"rating": "TotalRatingCount"})
)
```

```python
# movie_rating_count.head()
```

```python
final = data.merge(movie_rating_count, on="title", how="left")
```

```python
# final.head()
```

```python
final.to_parquet(path=r"./datasets/clean_data.parquet", engine="pyarrow", index=False)
```
